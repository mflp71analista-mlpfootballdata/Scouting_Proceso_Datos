```python
# Cargar los datos de la tabla jugadores que queremos analizar
query = "SELECT * \
         FROM jugadores \
         WHERE \"Competición\" = '2RFEF' AND \
               \"Posición Principal\" = 'Central' AND \
               \"Minutos jugados\" <= 300 AND \
               (\"Altura\" BETWEEN 180 AND 185 OR \"Altura\" = 0) AND \
               \"Edad\" BETWEEN 24 AND 27  \
        ;"
df = pd.read_sql(query, conexion_str)
df.drop(columns='Equipo', inplace=True)
df.rename(columns={'Equipo durante el período seleccionado': 'Equipo'}, inplace=True)
```

```python
# Aplicar las reglas y calcular la puntuación
#CENTRAL
df['Puntuacion'] = (
                    (df['Tiros interceptados/90'] > 1.0) * 1.7 +
                    (df['Interceptaciones/90'] > 4.5) * 1.4 +
                    (df['Posesión conquistada después de una interceptación'] > 8.5) * 1.3 +
                    (df['Acciones defensivas realizadas/90'] > 8.5) * 1.2 +
                    (df['Duelos aéreos/90'] > 4.5) * 1.1 +
                    (df['Duelos aéreos ganados %'] > 60) * 1.1
                    )

# Ordenar jugadores por puntuación
df = df.sort_values(by='Puntuacion', ascending=False)

df = df.drop_duplicates()

# Mostrar los 10 jugadores con mayor puntuación
df[['Jugador', 'Equipo', 'Minutos jugados', 'Edad', 'Altura', 'Puntuacion']]

# Guardar el resultado en una tabla nueva en la base de datos (opcional)
#df.to_sql('jugadores_priorizados', conexion_str, if_exists='replace', index=False)
```
